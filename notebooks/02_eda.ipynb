```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

correlation of pollutants with themselves

```python
df = pd.read_csv("../data/processed/X_train.csv")
# only picking the pollutants attributes ->
df = df[['PM2.5','PM10','NO','NO2','NOx','NH3','CO','SO2','O3','Benzene','Toluene']]
display(df)

corr = df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap='coolwarm',fmt=".2f")
plt.title("correlation between pollutants")
plt.show()
```

correlation of pollutants with AQI

```python
df1 = pd.read_csv("../data/processed/y_train.csv")
df['AQI'] = df1.values # combining the AQI column with the pollutants dataframe

corr = df.corr()
aqi_corr = corr['AQI'].sort_values(ascending=True)

plt.figure(figsize=(12,2))
sns.heatmap(aqi_corr.to_frame().T, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation of AQI with Each Pollutant")
plt.show()
```
